Cost of a wrong call: If a healthy page is ranked too highly, the team may waste limited review time. If an important declining page is ranked too low, the team may miss a useful improvement opportunity.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

he starter dataset contains 30,000 rows. About 54.21% of the pages are marked as declining, and the median 90-day impressions are 731. These numbers show that the dataset contains meaningful variation in visibility and performance, making page-priority scoring worth further investigation.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

total_rows = len(df)
declining_pct = (df["trend_direction"] == "down").mean() * 100
median_impressions = df["impressions_90d"].median()

print("Total rows:", total_rows)
print(f"Declining pages: {declining_pct:.2f}%")
print(f"Median 90-day impressions: {median_impressions:.0f}")
```
